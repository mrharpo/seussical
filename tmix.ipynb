```python
from sqlmodel import Session, select
from qlab.database_sqlmodel import CueDatabase
from qlab.models import Cue

DATABASE = 'mix/seuss15.tmix'
```

```python
with Session(CueDatabase(DATABASE).engine) as session:
    cues = list(session.exec(select(Cue).order_by(Cue.number)).all())
```

```python
from qlab import QLab

# q = QLab(address="192.168.1.17")
```

```python
import csv
```

```python
with open('sound-cues.csv', 'w') as f:
    writer = csv.writer(f)
    writer.writerow(['Page Number', 'Layer Title', 'Cue Number', 'Label', 'Work Note'])
    for cue in cues:
        writer.writerow(
            [
                0,
                'Sound',
                f'{cue.number}.{cue.point}' if cue.point else cue.number,
                cue.name,
            ]
        )
```
